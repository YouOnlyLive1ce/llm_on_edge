```python
import numpy as np
from scipy import stats
from scipy.stats import kruskal, mannwhitneyu
from itertools import combinations
import matplotlib.pyplot as plt

def comprehensive_statistical_test(y_data, group_labels, names, y_name):
    """
    Perform comprehensive statistical testing between groups
    
    Parameters:
    - y_data: array of values
    - group_labels: list of group assignments
    - names: group names
    - y_name: name of the metric being tested
    """
    
    # Group the data
    groups = {}
    for label, value in zip(group_labels, y_data):
        if label not in groups:
            groups[label] = []
        groups[label].append(value)
    
    group_values = [np.array(groups[i]) for i in range(len(names))]
    
    print(f"\n--- Statistical Analysis for {y_name} ---")
    print(f"Sample sizes: {[len(g) for g in group_values]}")
    
    # Check normality (Shapiro-Wilk test)
    print("\nNormality test (Shapiro-Wilk):")
    normal = True
    for i, group in enumerate(group_values):
        if len(group) >= 3:  # Shapiro requires n >= 3
            stat, p = stats.shapiro(group)
            is_normal = p > 0.05
            normal &= is_normal
            print(f"  {names[i]}: p={p:.4f} {'(normal)' if is_normal else '(non-normal)'}")
    
    # Choose appropriate test based on normality
    if normal:
        # Parametric tests
        f_stat, p_anova = stats.f_oneway(*group_values)
        print(f"\nOne-way ANOVA: F={f_stat:.3f}, p={p_anova:.4f}")
        
        if p_anova < 0.05:
            print("Significant differences found. Post-hoc tests:")
            
            # Check if statsmodels is available for Tukey HSD
            try:
                from statsmodels.stats.multicomp import pairwise_tukeyhsd
                tukey_results = pairwise_tukeyhsd(y_data, group_labels)
                print(tukey_results)
            except ImportError:
                # Fallback to pairwise t-tests with Bonferroni correction
                print("statsmodels not available, using pairwise t-tests with Bonferroni correction:")
                n_comparisons = len(names) * (len(names) - 1) // 2
                alpha_corrected = 0.05 / n_comparisons
                
                for (i, j) in combinations(range(len(names)), 2):
                    t_stat, p_val = stats.ttest_ind(group_values[i], group_values[j])
                    sig = "***" if p_val < alpha_corrected else ("*" if p_val < 0.05 else "ns")
                    print(f"  {names[i]} vs {names[j]}: t={t_stat:.3f}, p={p_val:.4f} {sig}")
        else:
            print(f"No significant differences between groups (p={p_anova})")
    else:
        # Non-parametric tests
        h_stat, p_kruskal = kruskal(*group_values)
        print(f"\nKruskal-Wallis H-test: H={h_stat:.3f}, p={p_kruskal:.4f}")
        
        if p_kruskal < 0.05:
            print("Significant differences found. Post-hoc Mann-Whitney U tests:")
            
            # Apply Bonferroni correction for multiple comparisons
            n_comparisons = len(names) * (len(names) - 1) // 2
            alpha_corrected = 0.05 / n_comparisons
            
            for (i, j) in combinations(range(len(names)), 2):
                u_stat, p_mw = mannwhitneyu(group_values[i], group_values[j], alternative='two-sided')
                sig = "***" if p_mw < alpha_corrected else ("*" if p_mw < 0.05 else "ns")
                print(f"  {names[i]} vs {names[j]}: U={u_stat:.0f}, p={p_mw:.4f} {sig}")
                print(f"    (Bonferroni corrected α = {alpha_corrected:.4f})")
        else:
            print("No significant differences between groups (p ≥ 0.05)")
    
    return group_values


def plot_tradeoffs_with_stats(x, *ys, group_labels=None, names=None, colors=None, 
                              titles=None, xlabel='base instability', 
                              figsize=(20, 4), ncols=4):
    """
    Plot tradeoffs and perform statistical tests between groups
    """
    nrows = 1
    fig, axes = plt.subplots(nrows, ncols, figsize=figsize)
    
    # Handle case where ncols=1
    if ncols == 1:
        axes = [axes]
    
    n_groups = len(names)
    
    # Store statistics for each y
    all_group_statistics = []
    
    for idx, y in enumerate(ys):
        if idx < len(axes):
            ax = axes[idx]
            
            # Collect data for each group
            group_statistics = []
            for group in range(n_groups):
                mask = [g == group for g in group_labels]
                group_data = np.array(y)[mask]
                group_statistics.append(group_data)
                
                # Plot individual group
                ax.scatter(np.array(x)[mask], group_data, 
                          label=names[group], 
                          color=colors[group] if colors else None,
                          alpha=0.7, s=50)
            
            all_group_statistics.append(group_statistics)
            
            ax.set_title(titles[idx] if titles else f'Y{idx+1}')
            ax.set_xlabel(xlabel)
            ax.legend(loc='best', fontsize='small')
            ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # Perform statistical tests
    print("\n" + "="*80)
    print("STATISTICAL ANALYSIS: Group Comparisons")
    print("="*80)
    
    comprehensive_statistical_test(
        y_data=np.array(x),
        group_labels=group_labels,
        names=names,
        y_name=["instability"]
    )
    
    for idx, (y, y_name) in enumerate(zip(ys, titles)):
        if idx < len(axes):
            # Perform statistical test
            comprehensive_statistical_test(
                y_data=np.array(y),
                group_labels=group_labels,
                names=names,
                y_name=y_name
            )
    
    return fig, axes, all_group_statistics


# ============================================================================
# USAGE EXAMPLE
# ============================================================================

# Your data
x = transitions['s base instability']
y1 = transitions['s base delta recall']
y2 = transitions['s base delta precision']
y3 = transitions['s base delta s_accuracy']
y4 = transitions['s base delta d_accuracy']
y5 = transitions['base ppl delta']

# Create the plot and perform statistical analysis
fig, axes, group_stats = plot_tradeoffs_with_stats(
    x,
    y1, y2, y3, y4, y5,
    group_labels=[i for i in [0,1,2,3,4]*5],
    names=["wanda 20%", "sparsegpt 20%", "no prune", "wanda 50%", "sparsegpt 50%"],
    colors=['green', 'blue', 'black', 'orange', 'red'],
    titles=['base delta recall', 'base delta precision', 'base delta s_accuracy', 
            'base delta d_accuracy', 'base ppl delta'],
    xlabel='base instability',
    ncols=5, 
    figsize=(20, 4)
)

# Optional: Access the grouped statistics for further analysis
print("\n\nGroup statistics summary:")
print("-" * 50)
for idx, (y_name, groups) in enumerate(zip(['base delta recall', 'base delta precision', 
                                            'base delta s_accuracy', 'base delta d_accuracy', 
                                            'base ppl delta'], group_stats)):
    print(f"\n{y_name}:")
    for name, group in zip(["wanda 20%", "sparsegpt 20%", "no prune", "wanda 50%", "sparsegpt 50%"], groups):
        print(f"  {name}: mean={np.mean(group):.4f}, std={np.std(group):.4f}, n={len(group)}")
```

```python
import json
import pandas as pd
from pathlib import Path

def convert_json_to_csv(json_files, output_folder=None, rows_per_file=10, max_rows=200):
    """
    Takes rows from each JSON file and combines into one CSV.
    
    Args:
        json_files: List of file paths
        output_folder: Path to folder for CSV output (default: current directory)
        rows_per_file: Number of rows to take from each file (default: 10)
        max_rows: Maximum total rows to process (default: 200)
    """
    if output_folder is None:
        output_path = Path.cwd()
    else:
        output_path = Path(output_folder)
        output_path.mkdir(parents=True, exist_ok=True)
    
    all_rows = []
    columns = ["model_hf_url", "user_question", "golden_answer", "answers", "category"]
    
    total_rows_processed = 0
    
    for file_path in json_files:
        # Stop if we've reached max rows
        if total_rows_processed >= max_rows:
            break
            
        file_path = Path(file_path)
        
        if not file_path.exists():
            print(f"File not found: {file_path}")
            continue
        
        try:
            # Read JSON file
            with open(file_path, 'r', encoding='utf-8') as f:
                data = json.load(f)
            
            # Convert to DataFrame
            if isinstance(data, list):
                df = pd.DataFrame(data)
            else:
                df = pd.DataFrame([data])
            
            # Calculate how many rows we can take
            remaining_rows = max_rows - total_rows_processed
            rows_to_take = min(rows_per_file, remaining_rows, len(df))
            
            if rows_to_take <= 0:
                break
                
            # Take rows using iloc with proper syntax (brackets, not parentheses)
            df_subset = df.iloc[total_rows_processed:total_rows_processed + rows_to_take].copy()
            
            # Check if all required columns exist, add missing ones as empty
            for col in columns:
                if col not in df_subset.columns:
                    df_subset[col] = ""
            
            # Select and reorder columns (only if columns exist)
            existing_columns = [col for col in columns if col in df_subset.columns]
            df_subset = df_subset[existing_columns]
            
            # Add source file info
            df_subset['source_file'] = file_path.name
            
            all_rows.append(df_subset)
            
            print(f"Took rows {total_rows_processed}-{total_rows_processed + rows_to_take} from {file_path.name}")
            
            total_rows_processed += rows_to_take
            
        except KeyError as e:
            print(f"Column error in {file_path.name}: {e}")
            # Try to handle missing columns
            try:
                # If columns missing, add them as empty
                for col in columns:
                    if col not in df.columns:
                        df[col] = ""
                df_subset = df.iloc[total_rows_processed:total_rows_processed + rows_to_take].copy()
                df_subset = df_subset[columns]
                df_subset['source_file'] = file_path.name
                all_rows.append(df_subset)
                print(f"Took rows {total_rows_processed}-{total_rows_processed + rows_to_take} from {file_path.name} (with missing columns filled)")
                total_rows_processed += rows_to_take
            except Exception as e2:
                print(f"Error processing {file_path.name} after column fix: {e2}")
        except Exception as e:
            print(f"Error processing {file_path.name}: {e}")
    
    if all_rows:
        # Combine all dataframes
        combined_df = pd.concat(all_rows, ignore_index=True)
        
        # Save to CSV
        csv_path = output_path / "combined_output.csv"
        combined_df.to_csv(csv_path, index=False, encoding='utf-8')
        
        print(f"\n✨ Conversion complete!")
        print(f"   Total rows: {len(combined_df)}")
        print(f"   Files processed: {len(all_rows)}")
        print(f"   Output saved to: {csv_path}")
    else:
        print("No data was processed.")

if __name__ == "__main__":
    json_files=[]
    import pandas as pd
    df=pd.read_csv("../../combined_output_human1.csv")
    for m in list(df["model_hf_url"].unique()):
        json_files.append(f"../../out/raw_results/{m[m.rfind("/")+1:]}.jsona")
            
    output_folder ="./"   # Set to a folder path if you want specific output location
    rows_per_file = 10    # Number of rows to take from each file
    max_rows = 200        # Maximum total rows to process
    
    convert_json_to_csv(json_files, output_folder, rows_per_file, max_rows)
```

```python
import os
import json
import platform
import psutil
from pathlib import Path
from typing import Dict, Any, Optional, List
from datetime import datetime

class ModelInfoExtractor:
    def __init__(self, model_path: str):
        self.model_path = Path(model_path)
        self.info = {}
        self.config = None
        
    def extract_all(self) -> Dict[str, Any]:
        """Extract all available information."""
        # Initialize with system info
        self.info = {
            "model_family": None,
            "exact_hf_model_id": None,
            "commit_hash": None,
            "base_vs_instruct": "instruct",
            "local_path": str(self.model_path.absolute()),
            "pruning_method": None,
            "sparsity": None,
            "quantization_method": None,
            "quantization_details": None,
            "backend": None,
            "backend_version": None,
            "device": None,
            "transformers_version": None,
            "cpu": "ryzen 5 5600",
            "cpu_arch": platform.processor(),
            "ram_gb": round(psutil.virtual_memory().total / (1024**3), 2),
            "os": f"{platform.system()} {platform.release()}",
            "thread_count": os.cpu_count(),
            "context_length": None,
            "model_architecture": None,
            "vocab_size": None,
            "num_layers": None,
            "hidden_size": None,
            "num_attention_heads": None,
            "batch_size": 1,
            "max_tokens": 64,
            "temperature": None,
            "repetition_penalty": None,
            "stop_sequence_token": None,
            "seed": None
        }
        
        # Check if it's a GGUF file or safetensors folder
        if self.model_path.suffix == '.gguf':
            self._extract_gguf_info()
        else:
            self._load_config()
            self._extract_from_config()
            self._get_transformers_version()
        
        return self.info
    
    def _extract_gguf_info(self):
        """Extract information from GGUF file."""
        try:
            # Try to import llama-cpp-python
            from llama_cpp import Llama
            
            # Load model to extract metadata
            llm = Llama(
                model_path=str(self.model_path),
                n_ctx=0,  # Don't load for inference, just metadata
                verbose=False
            )
            
            # Extract metadata
            metadata = llm.metadata
            
            # Model architecture
            if 'general.architecture' in metadata:
                arch = metadata['general.architecture']
                self.info["model_architecture"] = arch
                
                # Map to model family
                if 'llama' in arch.lower():
                    self.info["model_family"] = "Llama"
                elif 'olmo' in arch.lower():
                    self.info["model_family"] = "OLMo"
                elif 'mistral' in arch.lower():
                    self.info["model_family"] = "Mistral"
                elif 'qwen' in arch.lower():
                    self.info["model_family"] = "Qwen"
                else:
                    self.info["model_family"] = arch
            
            # Extract model parameters
            if f"{arch}.context_length" in metadata:
                self.info["context_length"] = int(metadata[f"{arch}.context_length"])
            if f"{arch}.vocab_size" in metadata:
                self.info["vocab_size"] = int(metadata[f"{arch}.vocab_size"])
            if f"{arch}.layer_count" in metadata:
                self.info["num_layers"] = int(metadata[f"{arch}.layer_count"])
            if f"{arch}.embedding_length" in metadata:
                self.info["hidden_size"] = int(metadata[f"{arch}.embedding_length"])
            if f"{arch}.attention.head_count" in metadata:
                self.info["num_attention_heads"] = int(metadata[f"{arch}.attention.head_count"])
            
            # Get HF model ID if available
            if 'general.name' in metadata:
                name = metadata['general.name']
                if 'olmo' in name.lower():
                    self.info["exact_hf_model_id"] = "allenai/OLMo-2-0425-1B-Instruct"
                elif 'smol' in name.lower():
                    self.info["exact_hf_model_id"] = "HuggingFaceTB/SmolLM2-1.7B-Instruct"
            
            # Extract quantization info from GGUF filename or metadata
            gguf_filename = self.model_path.name
            quant_patterns = {
                'IQ4_XS': 'IQ4_XS (4-bit, improved quantization)',
                'IQ4_NL': 'IQ4_NL (4-bit, no list)',
                'Q4_K_M': 'Q4_K_M (4-bit, medium)',
                'Q4_K_S': 'Q4_K_S (4-bit, small)',
                'Q5_K_M': 'Q5_K_M (5-bit, medium)',
                'Q5_K_S': 'Q5_K_S (5-bit, small)',
                'Q8_0': 'Q8_0 (8-bit)',
                'Q2_K': 'Q2_K (2-bit)',
                'Q3_K': 'Q3_K (3-bit)',
                'IQ1_S': 'IQ1_S (1-bit)',
                'IQ2_S': 'IQ2_S (2-bit)',
                'IQ3_S': 'IQ3_S (3-bit)'
            }
            
            for pattern, desc in quant_patterns.items():
                if pattern in gguf_filename:
                    self.info["quantization_method"] = desc
                    self.info["quantization_details"] = {
                        "format": "GGUF",
                        "quantization_type": pattern,
                        "description": desc
                    }
                    break
            else:
                # Try to get from metadata
                if 'general.file_type' in metadata:
                    self.info["quantization_method"] = f"GGUF (type {metadata['general.file_type']})"
            
            # Get token information
            if f"{arch}.tokenizer.eos_token_id" in metadata:
                self.info["stop_sequence_token"] = int(metadata[f"{arch}.tokenizer.eos_token_id"])
            
            # Set backend info
            self.info["backend"] = "llama.cpp"
            self.info["backend_version"] = self._get_llama_cpp_version()
            self.info["device"] = "cpu"  # or "cuda" if GPU is used
            
            # Close the model to free memory
            del llm
            
        except ImportError:
            # If llama-cpp-python is not installed, try basic file inspection
            self.info["backend"] = "llama.cpp"
            self.info["backend_version"] = "unknown (llama-cpp-python not installed)"
            self.info["device"] = "cpu"
            
            # Try to extract basic info from file size and name
            file_size_gb = self.model_path.stat().st_size / (1024**3)
            
            # Guess quantization from filename
            gguf_filename = self.model_path.name
            if 'IQ4_XS' in gguf_filename:
                self.info["quantization_method"] = "IQ4_XS (4-bit, improved quantization)"
                self.info["quantization_details"] = {"format": "GGUF", "quantization_type": "IQ4_XS"}
            elif 'Q4_K' in gguf_filename:
                self.info["quantization_method"] = "Q4_K (4-bit, K-quant)"
            elif 'Q5_K' in gguf_filename:
                self.info["quantization_method"] = "Q5_K (5-bit, K-quant)"
            elif 'Q8' in gguf_filename:
                self.info["quantization_method"] = "Q8_0 (8-bit)"
            
            print(f"Note: Install llama-cpp-python for full GGUF metadata extraction: pip install llama-cpp-python")
            print(f"File size: {file_size_gb:.2f} GB")
        
        # Set default context length if not found
        if not self.info["context_length"]:
            self.info["context_length"] = 4096  # Default for many models
    
    def _get_llama_cpp_version(self):
        """Get llama-cpp-python version."""
        try:
            import llama_cpp
            return llama_cpp.__version__
        except (ImportError, AttributeError):
            return "unknown"
    
    def _load_config(self):
        """Load config.json from model path."""
        config_path = self.model_path / "config.json"
        if config_path.exists():
            with open(config_path, 'r') as f:
                self.config = json.load(f)
        else:
            print(f"Warning: config.json not found in {self.model_path}")
    
    def _extract_from_config(self):
        """Extract all available information from config.json."""
        if not self.config:
            return
        
        # Model architecture/family
        architectures = self.config.get("architectures", [])
        if architectures:
            arch = architectures[0]
            self.info["model_architecture"] = arch
            # Extract model family
            if "Olmo2" in arch:
                self.info["model_family"] = "OLMo-2"
                self.info["exact_hf_model_id"] = "allenai/OLMo-2-0425-1B-Instruct"
            elif "Llama" in arch:
                self.info["model_family"] = "Llama"
                self.info["exact_hf_model_id"] = "HuggingFaceTB/SmolLM2-1.7B-Instruct"
            else:
                self.info["model_family"] = arch.replace("ForCausalLM", "").strip()
        
        # Check for pruning method from path
        local_path = str(self.info['local_path']).lower()
        if 'wanda' in local_path:
            self.info['pruning_method'] = 'wanda'
        elif 'sparsegpt' in local_path:
            self.info['pruning_method'] = 'sparsegpt'
        
        # Check for sparsity from path
        if '0.2' in local_path:
            self.info['sparsity'] = '0.2'
        elif '0.5' in local_path:
            self.info['sparsity'] = '0.5'
        
        # Model parameters
        self.info["context_length"] = self.config.get("max_position_embeddings")
        self.info["vocab_size"] = self.config.get("vocab_size")
        self.info["num_layers"] = self.config.get("num_hidden_layers")
        self.info["hidden_size"] = self.config.get("hidden_size")
        self.info["num_attention_heads"] = self.config.get("num_attention_heads")
        self.info["stop_sequence_token"] = self.config.get("eos_token_id")
        self.info["transformers_version"] = self.config.get("transformers_version")
        
        # Quantization (GPTQ specific)
        quant_config = self.config.get("quantization_config")
        if quant_config:
            self.info["quantization_method"] = quant_config.get("quant_method", "unknown")
            bits = quant_config.get("bits")
            group_size = quant_config.get("group_size")
            if bits and group_size:
                self.info["quantization_method"] = f"GPTQ (W{bits}, g{group_size})"
                self.info["quantization_details"] = {
                    "bits": bits,
                    "group_size": group_size,
                    "sym": quant_config.get("sym"),
                    "desc_act": quant_config.get("desc_act")
                }
            elif bits:
                self.info["quantization_method"] = f"GPTQ (W{bits})"
    
    def _get_transformers_version(self):
        """Get transformers and GPTQ backend versions."""
        try:
            import transformers
            self.info["transformers_version"] = transformers.__version__
        except ImportError:
            self.info["transformers_version"] = "unknown"
        
        # Check if it's a GPTQ model
        if not self.model_path.suffix == '.gguf' and self.config and self.config.get("quantization_config"):
            self.info["backend"] = "gptqmodel"
            try:
                meta = self.config.get("quantization_config", {}).get("meta", {})
                quantizer = meta.get("quantizer", [None])[0] if meta.get("quantizer") else None
                self.info["backend_version"] = quantizer or "unknown"
                self.info["device"] = meta.get("pack_impl", "cpu") if meta else "cpu"
            except:
                self.info["backend_version"] = "unknown"
                self.info["device"] = "cpu"
    
    def set_inference_params(self, 
                            batch_size: int = 1, 
                            max_tokens: int = 64,
                            temperature: float = None,
                            repetition_penalty: float = None,
                            seed: int = None):
        """Set inference-time parameters."""
        self.info["batch_size"] = batch_size
        self.info["max_tokens"] = max_tokens
        if self.info["backend"]=="llama.cpp":
            self.info["temperature"]=0.8
            self.info["repetition_penalty"]=1.0
        elif self.info["backend"]=="gptqmodel":
            self.info["temperature"]=0.0
            self.info["repetition_penalty"]=1.0
        if seed is not None:
            self.info["seed"] = seed

    def save_to_json(self, output_path: str):
        """Save information to JSON file."""
        Path(output_path).parent.mkdir(parents=True, exist_ok=True)
        with open(f"{output_path}.json", 'w') as f:
            json.dump(self.info, f, indent=2)
        print(f"Saved model summary to: {output_path}.json")
    
    def print_report(self):
        for key, value in self.info.items():
            if value is not None:
                # Format key for display
                display_key = key.replace('_', ' ').title()
                print(f"{display_key:30}: {value}")

# Example usage
if __name__ == "__main__":
    folder_path=Path('../../out')
    model_paths = models = [str(path) for path in folder_path.rglob("*.gguf")] + [str(path.parent) for path in folder_path.rglob("quantize_config.json")]
    
    for model_path in model_paths:
        if os.path.exists(model_path):
            print(f"\nProcessing: {model_path}")
            extractor = ModelInfoExtractor(model_path)
            info = extractor.extract_all()
            extractor.print_report()
            extractor.save_to_json(f"../../model_summaries/{Path(model_path).name}")
        else:
            print(f"Path not found: {model_path}")
```
